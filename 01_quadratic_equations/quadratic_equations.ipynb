# Quadratic Equations

A quadratic equation is an equation of the form:

$$ax^2 + bx + c = 0 \quad \text{with } a \neq 0$$

## The discriminant

To solve it, we compute the discriminant:

$$\Delta = b^2 - 4ac$$

The number of real solutions depends on the sign of $\Delta$:

- If $\Delta > 0$, the equation has **two distinct real roots**:

$$x_1 = \frac{-b - \sqrt{\Delta}}{2a} \quad \text{and} \quad x_2 = \frac{-b + \sqrt{\Delta}}{2a}$$

- If $\Delta = 0$, the equation has **one double root**:

$$x_0 = \frac{-b}{2a}$$

- If $\Delta < 0$, the equation has **no real root**.

## Objective

In this notebook, we implement the resolution of a quadratic equation in Python, test it on examples covering the three cases, and plot the corresponding parabolas to check the results graphically.

```python
import math

def solve_quadratic(a, b, c):
    delta = b ** 2 - (4 * a * c)
    if delta > 0:
        x1 = (- b - math.sqrt(delta)) / (2 * a)
        x2 = (- b + math.sqrt(delta)) / (2 * a)
        return x1, x2
    elif delta == 0:
        x0 = - b / (2 * a)
        return x0
    else:
        return "No Real Root"
```

```python
print(solve_quadratic(1, -3, 2))
```

```python
print(solve_quadratic(1, -4, 4))
```

```python
print(solve_quadratic(1, 1, 1))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 4, 200)
y = x ** 2 - 3 * x + 2

x1, x2 = solve_quadratic(1, -3, 2)

plt.plot(x, y)
plt.axhline(0, color="black")
plt.scatter([x1, x2], [0, 0], color="red")
plt.title("x² - 3x + 2")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-0.5, 4.5, 200)
y = x ** 2 - 4 * x + 4

x0 = solve_quadratic(1, -4, 4)

plt.plot(x, y)
plt.axhline(0, color="black")
plt.scatter([x0], [0], color="red")
plt.title("x² - 4x + 4")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 2, 200)
y = x ** 2 + x + 1

print(solve_quadratic(1, 1, 1))

plt.plot(x, y)
plt.axhline(0, color="black")
plt.title("x² + x + 1")
plt.show()
```

## Conclusion

The `solve_quadratic` function returns the expected results in all three cases:

| Equation | $\Delta$ | Result |
|---|---|---|
| $x^2 - 3x + 2 = 0$ | $1 > 0$ | two roots: $x_1 = 1$ and $x_2 = 2$ |
| $x^2 - 4x + 4 = 0$ | $0$ | one double root: $x_0 = 2$ |
| $x^2 + x + 1 = 0$ | $-3 < 0$ | no real root |

The plots confirm these results graphically:

- When $\Delta > 0$, the parabola crosses the x-axis at two points, which match the computed roots.
- When $\Delta = 0$, the parabola only touches the x-axis at its vertex, at the double root.
- When $\Delta < 0$, the parabola stays entirely above the x-axis and never meets it.

The algebraic resolution and the graphical reading therefore lead to the same conclusions.
